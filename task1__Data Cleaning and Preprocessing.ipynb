## Import library

```python
import pandas as pd
```

```python
#Load the dataset
df = pd.read_csv(r"C:\Users\LENOVO\Downloads\retail_store_sales.csv")
```

```python
#show dataset
print(df.head())
df
```

```python
# explore the dataset
print("Dataset shape:", df.shape)
print("\nData types:")
print(df.dtypes)
```

```python
# find missing value
print("\nMissing values per column:")
print(df.isnull().sum())
```

```python
# % of missing valuses
print("\nMissing Values (%)")
print((df.isnull().sum() / len(df) * 100).round(2))
```

```python
# check  duplicate rows
print("Full duplicate rows:", df.duplicated().sum())
df = df.drop_duplicates()
```

```python
#convert to lowercase and replace spaces with underscore
df.columns = df.columns.str.lower().str.replace(" ", "_")
print("\nColumns after renaming:")
print(df.columns.tolist())
```

```python
#convert date column to datetime
df["transaction_date"] = pd.to_datetime(df["transaction_date"])
print("\nDate column converted ")
print(df["transaction_date"].dtype)
```

```python
# Handle missing values
print("\nHandling Missing Values")

# for discount
df["discount_applied"] = df["discount_applied"].fillna(False)
df["discount_applied"] = df["discount_applied"].astype(bool)

# for item
df["item"] = df["item"].fillna("Unknown")

# recalculate Total Spent where possible
mask = df["price_per_unit"].notna() & df["quantity"].notna() & df["total_spent"].isna()
df.loc[mask, "total_spent"] = df.loc[mask, "price_per_unit"] * df.loc[mask, "quantity"]

# drop rows that still have missing important values
df = df.dropna(subset=["price_per_unit", "quantity", "total_spent"]).copy()

print("Shape after handling missing values:", df.shape)
```

```python
# drop rows where important numeric values are still missing
df = df.dropna(subset=["price_per_unit", "quantity", "total_spent"])
print("Shape after handling missing values:", df.shape)
```

```python
# fix data types
df["quantity"] = df["quantity"].astype(int)

print("\nData types after conversion:")
print(df.dtypes)
```

```python
#standardize text values
df["category"] = df["category"].str.strip().str.title()
df["item"] = df["item"].str.strip()
df["payment_method"] = df["payment_method"].str.strip().str.title()
df["location"] = df["location"].str.strip().str.title()

print("\nUnique values after standardization:")
print("Categories:", df["category"].unique())
print("Payment Methods:", df["payment_method"].unique())
print("Locations:", df["location"].unique())
```

```python
#Final check
print("\nFINAL CLEANED DATASET SUMMARY")
print("Final shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
print("\nData types:")
print(df.dtypes)
print("\nMissing values left:")
print(df.isnull().sum())
print("\nFirst 5 rows of cleaned data:")
print(df.head())
df
```

```python
#saved the cleaned dataset
df.to_csv("retail_store_sales_cleaned.csv", index=False)
print("\nCleaned dataset saved as: retail_store_sales_cleaned.csv")
```
